```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)

```

```python
df
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)

```

```python
df
```

```python
df['Ram'] = df['Ram'].str.replace('GB', '')
df['Ram'] = df['Ram'].astype(int)
```

```python
df
```

```python
df['Weight'] = df['Weight'].str.replace('kg', '')
df['Weight'] = df['Weight'].astype(float)
```

```python
df
```

```python
df['Price'] = df['Price'].astype(int)
```

```python
df
```

```python
#scatter plot
```

```python
sns.scatterplot(x='Company',y='Price',data=df,hue='Ram',size='Inches',palette='pastel')
```

```python
#line graph
```

```python
sns.lineplot(x='TypeName',y='Ram',data=df,hue='Company',palette='pastel',marker='o',style='Inches')
```

```python
#bar plot
```

```python
sns.barplot(x='Company',y='Price',data=df,hue='Company',palette='pastel')
```

```python
#histogram
```

```python
sns.histplot(x='Price',data=df,bins=10,hue='Company',palette='pastel',kde=True)
```

```python
#kde graph
```

```python
sns.kdeplot(x='Price',data=df,hue='Company',palette='pastel',fill=True)
```

```python
#box plot
```

```python
sns.boxplot(x='Company',y='Price',data=df,hue='Company',palette='pastel')
```

```python
#voiline plot
```

```python
sns.violinplot(x='Company',y='Price',data=df,hue='Company',palette='pastel')
```
